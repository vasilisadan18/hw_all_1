# **АНАЛИЗ ИСТОРИЧЕСКИХ ДАННЫХ**

```python
# Функция для анализа одного города (для того, чтобы использовать распараллеливание)
def analyze_city(city_data):
    city=city_data['city'].iloc[0]
    city_data=city_data.sort_values('timestamp')

    # Скользящее среднее за 30 дней
    city_data['rolling_mean']=city_data['temperature'].rolling(window=30, min_periods=1).mean()
    city_data['rolling_std']=city_data['temperature'].rolling(window=30, min_periods=1).std()

    # Границы для аномалий (среднее ± 2сигма)
    city_data['upper_limit'] = city_data['rolling_mean'] + 2*city_data['rolling_std']
    city_data['lower_limit'] = city_data['rolling_mean'] - 2*city_data['rolling_std']

    # Выявление аномалий
    city_data['is_anomaly'] = (city_data['temperature'] > city_data['upper_limit']) | \
                              (city_data['temperature'] < city_data['lower_limit'])
    # Тип аномалии (горячая/холодная)
    city_data['anomaly_type'] = np.where(
        city_data['is_anomaly'] & (city_data['temperature'] > city_data['upper_limit']),
        'hot',
        np.where(
            city_data['is_anomaly'] & (city_data['temperature'] < city_data['lower_limit']),
            'cold',
            'normal'
        )
    )
    # Отклонение от скользящего среднего
    city_data['deviation'] = city_data['temperature'] - city_data['rolling_mean']
    city_data['deviation_std'] = city_data['deviation'] / city_data['rolling_std'].replace(0, np.nan)

    # Cкользящее среднее за год
    city_data['rolling_mean_year'] = city_data['temperature'].rolling(window=365, min_periods=1).mean()

    return city_data
```

**Анализ *без распараллеливания*: **

```python
start_time1 = datetime.now()

results_no_parallel = []
for city in df['city'].unique():
    city_data = df[df['city'] == city].copy()
    analyzed_city = analyze_city(city_data)
    results_no_parallel.append(analyzed_city)

df_analyzed_no_parallel = pd.concat(results_no_parallel)
time_no_parallel = (datetime.now()-start_time1).total_seconds()
print(f"Время выполнения: {time_no_parallel:.2f} секунд")
```

 **Анализ *с распараллеливанием:* **

```python
start_time2 = datetime.now()

import multiprocessing as mp
from concurrent.futures import ProcessPoolExecutor, ThreadPoolExecutor

cities_list = [df[df['city'] == city] for city in df['city'].unique()]

with ProcessPoolExecutor(max_workers=mp.cpu_count()) as executor:
    results_parallel = list(executor.map(analyze_city, cities_list))

df_analyzed_parallel = pd.concat(results_parallel)
time_parallel = (datetime.now() - start_time2).total_seconds()
print(f"Время выполнения: {time_parallel:.2f} секунд")
```

Скорость анализа с распараллеливанием и без практически не отличается (без распараллеливания даже быстрее осуществляется). Видимо это связано с тем, что создание процессов требует времени. А данный датасет не очень большой и нет необходимости в распараллеливании вычислений.

***Анализ долгосрочных трендов:***

```python
def calculate_trend(city_data):
    """Рассчитывает долгосрочный тренд температуры"""
    if len(city_data) < 2:
        return 0

    # Преобразуем даты в числовой формат для регрессии
    x = np.arange(len(city_data))
    y = city_data['temperature'].values

    # Линейная регрессия
    coeffs=np.polyfit(x, y, 1)
    return coeffs[0]

# Рассчитываем тренды для каждого города
trends = {}
for city in df['city'].unique():
    city_data = df_analyzed_parallel[df_analyzed_parallel['city'] == city].sort_values('timestamp')
    trend = calculate_trend(city_data)
    trends[city]=trend

# Конвертируем в градусы/год
days_per_year = 365
trends_df = pd.DataFrame(list(trends.items()), columns=['city', 'trend_per_day'])
trends_df['trend_per_year'] = trends_df['trend_per_day']*days_per_year
trends_df = trends_df.sort_values('trend_per_year', ascending=False)

print("Города с наибольшим ростом температуры (градусы/год):")
print(trends_df.head())
print("Города с наибольшим снижением температуры (градусы/год):")
print(trends_df.tail())
```

***Синхронный способ получания текущей температуры:***

```python
def get_weather_sync(city, api_key):
    """Синхронный запрос температуры"""
    params = {
        'q': city,
        'appid': api_key,
        'units': 'metric'
    }
    try:
        response = requests.get(BASE_URL, params=params, timeout=10)
        if response.status_code == 200:
            data = response.json()
            return {
                'city': city,
                'temp': round(data['main']['temp'], 1),
                'success': True
            }
        return {'city': city, 'success': False, 'error': f'HTTP {response.status_code}'}
    except Exception as e:
        return {'city': city, 'success': False, 'error': str(e)}

sync_start = time.time()
sync_results = []

for city in CITIES_TO_CHECK:
    print(f"Запрос температуры для города {city} -", end=" ")
    result = get_weather_sync(city, API_KEY)
    sync_results.append(result)

    if result['success']:
        print(f"{result['temp']} градусов Цельсия")
    else:
        print(f"{result.get('error', 'Ошибка')}")

sync_time = time.time() - sync_start
print(f"Время синхронного метода: {sync_time:.2f} сек")

```

***Асинхронный запрос текущей погоды:***

```python
async def get_weather_async(session, city, api_key):
    """Асинхронный запрос температуры"""
    params = {
        'q': city,
        'appid': api_key,
        'units': 'metric'
    }

    try:
        async with session.get(BASE_URL, params=params, timeout=10) as response:
            if response.status == 200:
                data = await response.json()
                return {
                    'city': city,
                    'temp': round(data['main']['temp'], 1),
                    'success': True
                }
            return {'city': city, 'success': False, 'error': f'HTTP {response.status}'}
    except Exception as e:
        return {'city': city, 'success': False, 'error': str(e)}

async def get_all_weather_async(cities, api_key):
    """Асинхронные запросы для городов"""
    async with aiohttp.ClientSession() as session:
        tasks = [get_weather_async(session, city, api_key) for city in cities]
        return await asyncio.gather(*tasks)

async_start = time.time()

async def run_async():
    results = await get_all_weather_async(CITIES_TO_CHECK, API_KEY)
    async_time = time.time() - async_start

    for result in results:
        if result['success']:
            print(f"{result['city']}: {result['temp']} градусов Цельсия")
        else:
            print(f"{result['city']}: {result.get('error', 'Ошибка')}")

    print(f"Время асинхронного запроса: {async_time:.2f} сек")
    return results, async_time

async_results, async_time = asyncio.run(run_async())
```

Вывод: Использование асинхронного запроса текущей температуры затрачивает меньше времени. Получается, что асинхронный метод быстрее при множественных запросах.

***Анализ аномалий текущей погоды:***

```python
def get_current_season():
    """ Функция проверки текущего сезона"""
    month = datetime.now().month
    if month in [12, 1, 2]:
        return 'winter'
    elif month in [3, 4, 5]:
        return 'spring'
    elif month in [6, 7, 8]:
        return 'summer'
    else:
        return 'autumn'

current_season = get_current_season()
print(f"Текущий сезон {current_season}")
```

```python
def check_current_temp_anomaly(city, current_temp, current_season):
    """Проверяет аномальность текущей температуры используя season_stats"""

    # Находим статистику для города и сезона
    city_stats = season_stats[
        (season_stats['city'] == city) &
        (season_stats['season'] == current_season)
    ]

    if len(city_stats) == 0:
        return None

    stats = city_stats.iloc[0]
    mean_temp = stats['temperature_mean']
    std_temp = stats['temperature_std']

    # Рассчитываем границы как в analyze_city: среднее ± 2сигма
    upper_limit = mean_temp + 2 * std_temp
    lower_limit = mean_temp - 2 * std_temp

    # Проверяем аномальность
    if current_temp>upper_limit:
        return {
            'is_anomaly': True,
            'anomaly_type': 'hot',
            'deviation': current_temp - mean_temp,
            'upper_limit': upper_limit,
            'lower_limit': lower_limit,
            'mean': mean_temp,
            'std': std_temp
        }
    elif current_temp<lower_limit:
        return {
            'is_anomaly': True,
            'anomaly_type': 'cold',
            'deviation': current_temp - mean_temp,
            'upper_limit': upper_limit,
            'lower_limit': lower_limit,
            'mean': mean_temp,
            'std': std_temp
        }
    else:
        return {
            'is_anomaly': False,
            'anomaly_type': 'normal',
            'deviation': current_temp - mean_temp,
            'upper_limit': upper_limit,
            'lower_limit': lower_limit,
            'mean': mean_temp,
            'std': std_temp
        }
```

График временного ряда температур с аномалиями

```python
def plot_temperature_with_anomalies(city_data, city_name):
    fig, ax = plt.subplots(figsize=(14, 6))

    # Основной график температуры
    ax.plot(city_data['timestamp'], city_data['temperature'],
            label='Температура', color='blue', alpha=0.6, linewidth=1)

    # Скользящее среднее
    ax.plot(city_data['timestamp'], city_data['rolling_mean'],
            label='Скользящее среднее (30 дней)', color='orange', linewidth=2)

    # Аномалии-горячие
    if 'anomaly_type' in city_data.columns:
        hot_anomalies = city_data[city_data['anomaly_type'] == 'hot']
        if not hot_anomalies.empty:
            ax.scatter(hot_anomalies['timestamp'], hot_anomalies['temperature'],
                      color='red', s=30, label='Горячие аномалии', alpha=0.7)

    # Аномалии-холодные
    if 'anomaly_type' in city_data.columns:
        cold_anomalies = city_data[city_data['anomaly_type'] == 'cold']
        if not cold_anomalies.empty:
            ax.scatter(cold_anomalies['timestamp'], cold_anomalies['temperature'],
                      color='blue', s=30, label='Холодные аномалии', alpha=0.7)

    ax.set_title(f'Температурный ряд с аномалиями: {city_name}')
    ax.set_xlabel('Дата')
    ax.set_ylabel('Температура (C)')
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.xticks(rotation=45)
    plt.tight_layout()

    return fig

# Пример для Нью-Йорка:
city_name = 'New York'
city_data = df_analyzed[df_analyzed['city'] == city_name].sort_values('timestamp')
fig1 = plot_temperature_with_anomalies(city_data, city_name)
plt.show()
```

График сезонных температур:

```python
def plot_seasonal_profiles(city_data, city_name):
    seasons = ['winter', 'spring', 'summer', 'autumn']
    season_names= ['Зима', 'Весна', 'Лето', 'Осень']
    colors= ['blue', 'green', 'red', 'orange']
    fig, ax = plt.subplots(figsize=(10, 6))
    x_pos = np.arange(len(seasons))
    means=[]
    stds=[]

    for season in seasons:
        season_data= city_data[city_data['season'] == season]
        if not season_data.empty:
            means.append(season_data['temperature'].mean())
            stds.append(season_data['temperature'].std())
        else:
            means.append(0)
            stds.append(0)

    bars =ax.bar(x_pos, means, color=colors, alpha=0.7)
    ax.errorbar(x_pos, means, yerr=stds, fmt='none',
                ecolor='black', capsize=5, capthick=2)

    for i, (mean, std) in enumerate(zip(means, stds)):
        ax.text(i, mean + std + 0.5, f'{mean:.1f}+-{std:.1f} C',
                ha='center', va='bottom', fontsize=10)

    ax.set_title(f'Сезонные профили температуры: {city_name}')
    ax.set_xlabel('Сезон')
    ax.set_ylabel('Температура ( C)')
    ax.set_xticks(x_pos)
    ax.set_xticklabels(season_names)
    ax.grid(True, alpha=0.3, axis='y')
    plt.tight_layout()

    return fig

city_name= 'New York'
city_data= df_analyzed[df_analyzed['city'] == city_name].sort_values('timestamp')
fig2 =plot_seasonal_profiles(city_data, city_name)
plt.show()
```
